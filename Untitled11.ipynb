```python
# ── Cell 5.1: Patch benchmark script ─────────────────────────────────

from pathlib import Path
import re

script_path = Path("asr_benchmark.py")

text = script_path.read_text()

# ─────────────────────────────────────────────────────────────
# Better normalization
# ─────────────────────────────────────────────────────────────

old_normalise = r'''
def normalise(text: str) -> str:
    """Lowercase, transliterate Devanagari to Roman, strip punctuation."""
    text = transliterate_to_roman(text)
    text = text.lower()
    text = re.sub(r'[^\w\s]', ' ', text, flags=re.UNICODE)
    text = re.sub(r'\s+', ' ', text).strip()
    return text
'''

new_normalise = r'''
import unicodedata

def normalise(text: str) -> str:

    text = transliterate_to_roman(text)

    text = unicodedata.normalize("NFKD", text)
    text = text.encode("ascii", "ignore").decode("ascii")

    text = text.lower()

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    return text
'''

text = text.replace(old_normalise, new_normalise)

# ─────────────────────────────────────────────────────────────
# Better entity matching
# ─────────────────────────────────────────────────────────────

old_check = r'''
def check_entity(hyp: str, entity: str) -> Tuple[bool, bool]:
    """
    Returns (exact_hit, fuzzy_hit).
    exact_hit  — entity found after full normalisation (script-agnostic)
    fuzzy_hit  — any known Devanagari variant found in raw hypothesis
    """
    hyp_norm = normalise(hyp)
    ent_norm = normalise(entity)

    # Exact: all words of entity appear in order in hypothesis
    tokens = ent_norm.split()
    pattern = r'.*'.join(re.escape(t) for t in tokens)
    exact = bool(re.search(pattern, hyp_norm))

    # Fuzzy: check raw Devanagari variants
    fuzzy = exact
    if not fuzzy:
        variants = LOCALITY_DEVANAGARI.get(entity, [])
        for v in variants:
            if v in hyp:
                fuzzy = True
                break

    return exact, fuzzy
'''

new_check = r'''
from rapidfuzz import fuzz

def check_entity(hyp: str, entity: str) -> Tuple[bool, bool]:

    hyp_norm = normalise(hyp)
    ent_norm = normalise(entity)

    score = fuzz.partial_ratio(ent_norm, hyp_norm)

    exact = score >= 85
    fuzzy = score >= 70

    return exact, fuzzy
'''

text = text.replace(old_check, new_check)

# ─────────────────────────────────────────────────────────────
# Whisper improvements
# ─────────────────────────────────────────────────────────────

pattern_whisper = r'''
result = _whisper_model\.transcribe\(
\s*path,
.*?
\s*verbose=False,
.*?
\)
'''

replacement_whisper = '''
result = _whisper_model.transcribe(
        path,
        task="transcribe",
        language="hi",
        fp16=False,
        verbose=False,
        condition_on_previous_text=False,
        temperature=0,
    )
'''

text = re.sub(
    pattern_whisper,
    replacement_whisper,
    text,
    flags=re.DOTALL
)

# ─────────────────────────────────────────────────────────────
# Faster Whisper improvements
# ─────────────────────────────────────────────────────────────

pattern_fw = r'''
segments, info = _fw_model\.transcribe\(
\s*path,
.*?
\s*vad_parameters=\{"min_silence_duration_ms": 300\},
.*?
\)
'''

replacement_fw = '''
segments, info = _fw_model.transcribe(
        path,
        language="hi",
        beam_size=5,
        vad_filter=True,
        vad_parameters={"min_silence_duration_ms": 300},
        condition_on_previous_text=False,
        temperature=0,
    )
'''

text = re.sub(
    pattern_fw,
    replacement_fw,
    text,
    flags=re.DOTALL
)

# ─────────────────────────────────────────────────────────────
# Increase Deepgram timeout
# ─────────────────────────────────────────────────────────────

text = text.replace(
    'timeout=60,',
    'timeout=120,'
)

# ─────────────────────────────────────────────────────────────
# Save CSV correctly to Drive
# ─────────────────────────────────────────────────────────────

old_save = r'''
def save_results(df: pd.DataFrame, output_path: Path) -> None:
    base = Path(output_path).stem
    df.to_csv(f"{base}.csv", index=False)
    df.to_json(str(output_path), orient="records", indent=2)
    print(f"\nResults saved → {base}.csv  |  {output_path}")
'''

new_save = r'''
def save_results(df: pd.DataFrame, output_path: Path) -> None:

    output_path = Path(output_path)

    csv_path = output_path.with_suffix(".csv")

    df.to_csv(csv_path, index=False)

    df.to_json(str(output_path), orient="records", indent=2)

    print(f"\nResults saved → {csv_path} | {output_path}")
'''

text = text.replace(old_save, new_save)

# ─────────────────────────────────────────────────────────────
# Write patched script
# ─────────────────────────────────────────────────────────────

script_path.write_text(text)

print("✅ Benchmark script patched successfully")
```

```python
# ── Final Run Cell ───────────────────────────────────────────────────

import importlib.util
import pathlib

spec = importlib.util.spec_from_file_location(
    'asr_benchmark',
    'asr_benchmark.py'
)

bm = importlib.util.module_from_spec(spec)

spec.loader.exec_module(bm)

# ─────────────────────────────────────────────────────────────

class Args:

    audio_dir = pathlib.Path(AUDIO_DIR)

    models = MODELS

    whisper_size = WHISPER_SIZE

    output = pathlib.Path(OUTPUT_JSON)

    charts = True

    charts_dir = pathlib.Path(CHARTS_DIR)

args = Args()

print('🚀 Loading models...')

runners = bm.build_runners(
    MODELS.split(','),
    args
)

print('\n🎙️ Running benchmark...')

df = bm.run_benchmark(
    args.audio_dir,
    runners
)

print('\n📊 Summary')

bm.print_summary(df)

print('\n💾 Saving results...')

bm.save_results(
    df,
    args.output
)

print('\n📈 Generating charts...')

bm.generate_charts(
    df,
    args.charts_dir
)

print('\n✅ BENCHMARK COMPLETE')

print('\n📁 Results saved to:')
print(' -', OUTPUT_JSON)
print(' -', CHARTS_DIR)
```

```python
# ── NEW CELL: Model Size Tradeoff Experiment ─────────────────────────

import time
import pathlib
import pandas as pd

from faster_whisper import WhisperModel

# ─────────────────────────────────────────────────────────────
# TEST CONFIG
# ─────────────────────────────────────────────────────────────

TEST_MODELS = [
    'small',
    'medium',
    'large-v3',
]

TEST_FILES = [
    '01_koramangala_quiet.wav',
    '02_indiranagar_street.wav',
    '04_electroniccity_rushed.wav',
    '08_hebbal_whispered.wav',
    '27_thanisandra_phone.wav',
]

audio_dir = pathlib.Path(AUDIO_DIR)

summary_rows = []

# ─────────────────────────────────────────────────────────────
# RUN EXPERIMENT
# ─────────────────────────────────────────────────────────────

for model_size in TEST_MODELS:

    print("\n" + "="*70)
    print(f"🚀 Testing Faster Whisper: {model_size}")
    print("="*70)

    try:

        # ─────────────────────────────────────────────────────
        # LOAD MODEL ON CPU
        # ─────────────────────────────────────────────────────

        model = WhisperModel(
            model_size,
            device="cpu",
            compute_type="int8"
        )

        rows = []

        # ─────────────────────────────────────────────────────
        # RUN FILES
        # ─────────────────────────────────────────────────────

        for filename in TEST_FILES:

            path = audio_dir / filename

            if not path.exists():

                continue

            reference = bm.GROUND_TRUTH.get(filename)

            locality = filename.split('_')[1]

            if reference is None:

                continue

            try:

                start = time.time()

                segments, info = model.transcribe(
                    str(path),
                    language="hi",
                    beam_size=5,
                    vad_filter=True,
                    condition_on_previous_text=False,
                    temperature=0,
                )

                latency = time.time() - start

                hyp = " ".join(
                    s.text for s in segments
                ).strip()

                # ─────────────────────────────────────────
                # NORMALIZE
                # ─────────────────────────────────────────

                ref_norm = bm.normalise(reference)

                hyp_norm = bm.normalise(hyp)

                # ─────────────────────────────────────────
                # METRICS
                # ─────────────────────────────────────────

                wer = bm.compute_wer(
                    ref_norm,
                    hyp_norm
                )

                exact, fuzzy = bm.check_entity(
                    hyp,
                    locality
                )

                rows.append({
                    'file': filename,
                    'wer': wer,
                    'latency': latency,
                    'entity_hit': int(exact),
                    'entity_fuzzy': int(fuzzy),
                })

                print(f"✅ {filename}")

                print(f"   → {hyp}")

            except Exception as e:

                print(f"❌ {filename}: {e}")

        # ─────────────────────────────────────────────────
        # SUMMARY
        # ─────────────────────────────────────────────────

        df = pd.DataFrame(rows)

        if len(df) == 0:

            continue

        summary_rows.append({
            'model_size': model_size,
            'avg_wer': round(df['wer'].mean(), 3),
            'avg_latency': round(df['latency'].mean(), 3),
            'entity_hit': round(df['entity_hit'].mean(), 3),
            'entity_fuzzy': round(df['entity_fuzzy'].mean(), 3),
        })

    except Exception as e:

        print(f"❌ Failed loading model {model_size}: {e}")

# ─────────────────────────────────────────────────────────────
# FINAL RESULTS
# ─────────────────────────────────────────────────────────────

summary_df = pd.DataFrame(summary_rows)

print("\n")
print("="*70)
print("📊 MODEL SIZE TRADEOFF RESULTS")
print("="*70)

display(summary_df)

# ─────────────────────────────────────────────────────────────
# SAVE
# ─────────────────────────────────────────────────────────────

save_path = (
    '/content/drive/MyDrive/vahan_asr/'
    'model_size_experiment.csv'
)

summary_df.to_csv(
    save_path,
    index=False
)

print("\n✅ Saved:")
print(save_path)
```

```python
# ── NEW CELL: Experiment 2 — Condition-wise Analysis ─────────────────
# Analyze which recording condition hurts ASR most

import pandas as pd

# ─────────────────────────────────────────────────────────────
# LOAD RESULTS
# ─────────────────────────────────────────────────────────────

results_path = '/content/drive/MyDrive/vahan_asr/results.csv'

df = pd.read_csv(results_path)

# ─────────────────────────────────────────────────────────────
# CONDITION ANALYSIS
# ─────────────────────────────────────────────────────────────

condition_summary = (
    df.groupby(['condition', 'model'])
      .agg({
          'wer': 'mean',
          'entity_hit': 'mean',
          'latency_s': 'mean',
      })
      .reset_index()
)

condition_summary = condition_summary.sort_values(
    by=['condition', 'wer']
)

print("\n")
print("="*70)
print("📊 CONDITION-WISE ANALYSIS")
print("="*70)

display(condition_summary)

# ─────────────────────────────────────────────────────────────
# WORST CONDITIONS
# ─────────────────────────────────────────────────────────────

worst = (
    condition_summary.groupby('condition')['wer']
    .mean()
    .sort_values(ascending=False)
)

print("\n")
print("="*70)
print("🔥 HARDEST AUDIO CONDITIONS")
print("="*70)

display(worst)

# ─────────────────────────────────────────────────────────────
# SAVE
# ─────────────────────────────────────────────────────────────

condition_summary.to_csv(
    '/content/drive/MyDrive/vahan_asr/condition_analysis.csv',
    index=False
)

print("\n✅ Saved:")
print("/content/drive/MyDrive/vahan_asr/condition_analysis.csv")
```

```python
# ── NEW CELL: Generate Better Charts ─────────────────────────────────

import pandas as pd
import matplotlib.pyplot as plt

# ─────────────────────────────────────────────────────────────
# LOAD RESULTS
# ─────────────────────────────────────────────────────────────

df = pd.read_csv(
    '/content/drive/MyDrive/vahan_asr/results.csv'
)

# ─────────────────────────────────────────────────────────────
# OUTPUT DIR
# ─────────────────────────────────────────────────────────────

chart_dir = '/content/drive/MyDrive/vahan_asr/charts'

# ─────────────────────────────────────────────────────────────
# CHART 1 — WER BY MODEL
# ─────────────────────────────────────────────────────────────

wer = (
    df.groupby('model')['wer']
      .mean()
)

plt.figure(figsize=(7,5))

wer.plot(kind='bar')

plt.title('Average WER by Model')
plt.ylabel('WER')
plt.tight_layout()

plt.savefig(f'{chart_dir}/wer_by_model.png')

plt.close()

# ─────────────────────────────────────────────────────────────
# CHART 2 — ENTITY HIT
# ─────────────────────────────────────────────────────────────

entity = (
    df.groupby('model')['entity_hit']
      .mean()
)

plt.figure(figsize=(7,5))

entity.plot(kind='bar')

plt.title('Entity Hit Rate by Model')
plt.ylabel('Entity Hit')

plt.tight_layout()

plt.savefig(f'{chart_dir}/entity_hit_by_model.png')

plt.close()

# ─────────────────────────────────────────────────────────────
# CHART 3 — LATENCY VS ACCURACY
# ─────────────────────────────────────────────────────────────

summary = (
    df.groupby('model')
      .agg({
          'wer': 'mean',
          'latency_s': 'mean',
      })
      .reset_index()
)

plt.figure(figsize=(7,5))

plt.scatter(
    summary['latency_s'],
    summary['wer']
)

for _, row in summary.iterrows():

    plt.text(
        row['latency_s'],
        row['wer'],
        row['model']
    )

plt.xlabel('Latency (s)')
plt.ylabel('WER')

plt.title('Latency vs Accuracy')

plt.tight_layout()

plt.savefig(f'{chart_dir}/latency_vs_accuracy.png')

plt.close()

print("✅ Charts generated")
print(f"📁 Saved to: {chart_dir}")
```
